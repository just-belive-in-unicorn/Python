1. (6б). Створіть клас Building, додайте три атрибути, один з яких має значення за замовчуванням та два методи на свій розсуд.

```python
class Building:

    # Constructor with 1 default value
    def __init__(self, height, width, length=100):
        self.height = height
        self.width = width
        self.length = length

    # Destroys the building
    def destroy(self):
        self.height = 0
        self.width = 0
        self.length = 0
        print(
            f"Building is now destroyed. It's length is {self.length}, it's width is {self.width}, it's height is {self.height}")

    # Destroys 1 floor
    def destroy_1_floor(self, floor_height):
        self.height = self.height - floor_height
        print(f"Destroyed 1 floor. Now height of the building is {self.height} m.")
```

5. (9б). Створіть клас 3D-фігура без атрибутів, з методом get_volume для отримання об'єму фігури, що повертає 0 та __add __, який приймає self та other в якості аргументів, а повертає суму об'ємів 3d-фігур self та other.

```python
import math


class Shape3D:

    def get_volume(self):
        return 0

    def __add__(self, other):
        if isinstance(other, Shape3D):
            return self.get_volume() + other.get_volume()
        else:
            return 0


shape1 = Shape3D()
shape2 = Shape3D()

result = shape1 + shape2
print(result)
```

6. (11б). Створіть 2 дочірніх класи від 3d-фігура: Циліндр та Куб, які мають атрибути, необхідні для розрахунку об'єму. Визначте метод get_volume в кожному з них так, щоби вони розраховували об'єм у залежності від формули для кожного типу фігури. Створіть обʼєкт класу Циліндр та обʼєкт класу Куб. Виконайте операцію суми за допомогою оператора + між ними, аби знайти сумарний об'єм.

```python
class Cylinder(Shape3D):
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def get_volume(self):
        return math.pi * (self.radius ** 2) * self.height


class Cube(Shape3D):
    def __init__(self, side_length):
        self.side_length = side_length

    def get_volume(self):
        return self.side_length ** 3


cylinder = Cylinder(1, 2)
cube = Cube(3)

print(f"Cylinder volume: {cylinder.get_volume():.2f}")
print(f"Cube volume: {cube.get_volume():.2f}")

total_volume = cylinder + cube
print(f"Total volume: {total_volume:.2f}")
```

7. (3б). Продемонструйте різницю між isinstance та issubclass.

```python
# isinstance checks if an object is an instance of a class
print(isinstance(5, int))          # True, because 5 is an int
print(isinstance(5.0, int))        # False, because 5.0 is a float, not an int
print(isinstance(5, (int, float))) # True, because 5 is an int, which is in the tuple

# issubclass checks inheritance relationships
print(issubclass(bool, int))       # True, because bool is a subclass of int
print(issubclass(int, float))      # False, because int is not a subclass of float
print(issubclass(bool, object))    # True, because all classes inherit from object
```

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

```python
class EWallet:
    def __init__(self, id, initial_balance=0):
        self.__id = id
        self.__balance = initial_balance

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, value):
        self.__balance = value

    @property
    def id(self):
        return self.__id

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive")
        self.balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive")
        if amount > self.balance:
            raise ValueError("Insufficient funds")
        self.balance -= amount


wallet = EWallet(id="1234", initial_balance=100)

print(f"Initial balance: {wallet.balance}")

wallet.deposit(50)
print(f"Balance after deposit: {wallet.balance}")

wallet.withdraw(10)
print(f"Balance after withdrawal: {wallet.balance}")

try:
    wallet.deposit(-10)
except ValueError as e:
    print(e)
```

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

```python
class Item:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __repr__(self):
        return f"Item(name={self.name}, price={self.price})"


class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, item):
        if isinstance(item, Item):
            self.items.append(item)
        else:
            raise ValueError("Only items of type 'Item' can be added to the cart")

    def remove_item(self, item):
        if item in self.items:
            self.items.remove(item)
        else:
            raise ValueError("Item not found in the cart")

    def display_items(self):
        if not self.items:
            print("Your shopping cart is empty.")
        else:
            for item in self.items:
                print(item)


cart = ShoppingCart()

item1 = Item("Cheese", 10)
item2 = Item("Milk", 20)
item3 = Item("Cookies", 30)

cart.add_item(item1)
cart.add_item(item2)
cart.add_item(item3)

print("Added items: ")
cart.display_items()
cart.remove_item(item2)
print("After removing item2: ")
cart.display_items()
```

Створіть клас Car, який успадковує обидва класи Vehicle та Engine. Продемонструйте використання множинної спадковості, створивши обʼєкт класу Car та отримавши mro для цього класу.

```python
class Vehicle:
    def __init__(self, brand, production_age):
        self.brand = brand
        self.production_age = production_age

    def __repr__(self):
        return f"Vehicle(brand={self.brand}, production_age={self.production_age})"


class Engine:
    def __init__(self, engine_type, size):
        self.engine_type = engine_type
        self.size = size

    def __repr__(self):
        return f"Engine(type={self.engine_type}, size={self.size}L)"


class Car(Vehicle, Engine):
    def __init__(self, brand, production_age, engine_type, size):
        Vehicle.__init__(self, brand, production_age)
        Engine.__init__(self, engine_type, size)

    def __repr__(self):
        return f"Car(brand={self.brand}, production_age={self.production_age}, engine_type={self.engine_type}, engine_size={self.size}L)"


car = Car(brand="Toyota", production_age=15, engine_type="Petrol", size=3.5)

print(car)
print("MRO for Car:", Car.mro())
```

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

```python
class WeekDaysIterator:
    def __init__(self):
        self.days_of_week = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.days_of_week):
            day = self.days_of_week[self.index]
            self.index += 1
            return day
        else:
            raise StopIteration


week_iterator = WeekDaysIterator()

for day in week_iterator:
    print(day)
```
